```python

import google.generativeai as genai

# Cấu hình API key
genai.configure(api_key="API_KEY")

# Prompt hệ thống (system instruction)
SYSTEM_PROMPT = """
Bạn là một chuyên gia về phân cấp tài liệu.

Hiện tại tài liệu của tôi ở dạng Markdown nhưng chưa được phân cấp rõ ràng.
Bây giờ tôi cần bạn giúp tôi phân cấp lại tài liệu này theo cấu trúc rõ ràng với các đề mục lớn, đề mục nhỏ và đề mục con.
Yêu cầu cụ thể:
Trả về nội dung dưới dạng **Markdown có phân cấp rõ ràng**:
   - Đề mục lớn (chương, bài): `#`
   - Đề mục trung bình (mục nhỏ): `##`
   - Đề mục con (tiểu mục, ví dụ 1.1.1): `###`
   - Cứ thêm các # với các đề mục nhỏ hơn nữa.
   - Giữ nguyên đánh số hoặc ký hiệu gốc của tiêu đề.


📘 Đầu ra cuối cùng:
- Văn bản Markdown có cấu trúc rõ ràng.
"""




print("✅ Model OCR đã sẵn sàng.")
```

```python
from PIL import Image
import fitz
import os, time

PDF_PATH = "CanhDieu/SGK TIN HOC 12 CS - CANH DIEU.pdf"

doc = fitz.open(PDF_PATH)

model = genai.GenerativeModel(
    model_name="gemini-2.5-flash",
    system_instruction=SYSTEM_PROMPT
)

start_page, end_page = 6, 163 
output_path = "SGK_Tin12_CD.md"
os.makedirs("temp_images", exist_ok=True)

with open(output_path, "a", encoding="utf-8") as f_out:  # a = append
    for PAGE_INDEX in range(start_page, end_page):
        page = doc.load_page(PAGE_INDEX)
        pix = page.get_pixmap(dpi=200)
        img_path = f"temp_images/page_{PAGE_INDEX+1}.png"
        pix.save(img_path)

        image = Image.open(img_path)
        response = model.generate_content([
            f"Trích xuất văn bản tiếng Việt từ hình ảnh trang {PAGE_INDEX+1}."
            , image
        ])

        page_text = ""
        if hasattr(response, "parts") and response.parts:
            for part in response.parts:
                if hasattr(part, "text"):
                    page_text += part.text.strip() + "\n"

        f_out.write(f"# --- Trang {PAGE_INDEX+1} ---\n\n")
        f_out.write(page_text + "\n")

        print(f"✅ Đã OCR xong trang {PAGE_INDEX+1}")

print(f"🎉 Đã nối tiếp nội dung vào → {output_path}")
```

```python
import google.generativeai as genai

input_path = "SGK_Tin12_CD_clean.md"
with open(input_path, "r", encoding="utf-8") as f:
        raw_markdown = f.read()

import re

def chunk_markdown(text, max_chars=9000):
    sentences = re.split(r'(?<=[.!?])\s+', text)
    chunks = []
    current = ""
    
    for s in sentences:
        if len(current) + len(s) + 1 > max_chars:
            chunks.append(current.strip())
            current = s
        else:
            current += " " + s

    if current.strip():
        chunks.append(current.strip())

    return chunks


import google.generativeai as genai

def process_chunk(chunk_text, SYSTEM_PROMPT, model_name="gemini-2.0-flash"):
    model = genai.GenerativeModel(
        model_name=model_name,
        system_instruction=SYSTEM_PROMPT
    )

    response = model.generate_content([
        "Hãy phân cấp lại đoạn markdown sau:",
        chunk_text
    ])

    result = ""
    if hasattr(response, "parts") and response.parts:
        for part in response.parts:
            if hasattr(part, "text"):
                result += part.text

    return result.strip()

def restructure_markdown_chunked(md_text, SYSTEM_PROMPT,
                                 max_chars=9000,
                                 model_name="gemini-2.0-flash"):

    chunks = chunk_markdown(md_text, max_chars=max_chars)
    print(f"🔹 Tổng số chunk: {len(chunks)}")

    final_output = []

    for i, chunk in enumerate(chunks):
        print(f"⚙️ Đang xử lý chunk {i+1}/{len(chunks)}...")

        processed = process_chunk(
            chunk_text=chunk,
            SYSTEM_PROMPT=SYSTEM_PROMPT,
            model_name=model_name
        )

        final_output.append(processed)

    print("🎉 Hoàn tất ghép tài liệu.")
    return "\n\n".join(final_output)


clean_markdown = restructure_markdown_chunked(
    md_text=raw_markdown,
    SYSTEM_PROMPT=SYSTEM_PROMPT,
    max_chars=9000
)

with open("SGK_Tin12_CD_clean.md", "w", encoding="utf-8") as f:
    f.write(clean_markdown)
```
